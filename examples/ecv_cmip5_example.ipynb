# ECV CMIP5 — sis-ecv-cmip5-bias-corrected

This notebook demonstrates how to query the **Essential Climate Variables CMIP5
bias-corrected** dataset (`sis-ecv-cmip5-bias-corrected`) via the Polytope DSS API.

The dataset provides bias-corrected climate projections derived from CMIP5 model
simulations. Both **bounding-box** and **polygon** spatial queries are shown.

The response is a [CoverageJSON](https://covjson.org) `CoverageCollection` that can be
decoded with [`covjsonkit`](https://github.com/ecmwf/covjsonkit) and converted to an
`xarray.Dataset`.

## Bounding-box query

```python
import requests

url = "https://polytope-dss.ecmwf.int/api/v1/dataset/sis-ecv-cmip5-bias-corrected/extract/?download_file=test.zip"
headers = {
    "accept": "application/json",
    "Content-Type": "application/json",
}
data = {
    "request": {
        "variable": ["mean_2m_temperature"],
        "period": ["20060101_20301231"],
    },
    "feature": {
        "type": "boundingbox",
        "coordinates": [[49.5, 5.5], [50.5, 6.5]],
    },
}

response = requests.post(url, headers=headers, json=data)

# Print the response
print(response.status_code)
print(response.json())
```

```python
from covjsonkit.api import Covjsonkit
decoder = Covjsonkit().decode(response.json())

ds = decoder.to_xarray()
ds
```

## Polygon query

```python
import requests

url = "https://polytope-dss.ecmwf.int/api/v1/dataset/sis-ecv-cmip5-bias-corrected/extract/?download_file=test.zip"
headers = {
    "accept": "application/json",
    "Content-Type": "application/json",
}
data = {
    "request": {
        "variable": ["mean_2m_temperature"],
        "period": ["20060101_20301231"],
    },
    "feature": {
        "type": "polygon",
        "coordinates": [
            [49.5, 5.5],
            [50.0, 5.5],
            [50.0, 6.0],
            [49.5, 6.0],
            [49.5, 5.5],
        ],
    },
}

response = requests.post(url, headers=headers, json=data)

# Print the response
print(response.status_code)
print(response.json())
```

```python
from covjsonkit.api import Covjsonkit
decoder = Covjsonkit().decode(response.json())

ds = decoder.to_xarray()
ds
```
