# Hydrology Variables — sis-hydrology-variables-derived-projections

This dataset provides hydrological variables derived from climate projections for European river basins. It includes variables such as river discharge under different climate scenarios and time periods. The API returns data in CoverageJSON format (`CoverageCollection`), which can be decoded and converted to an xarray Dataset using `covjsonkit`.

## Bounding-box query

```python
import requests

url = "https://polytope-dss.ecmwf.int/api/v1/dataset/sis-hydrology-variables-derived-projections/extract/?download_file=test.zip"
headers = {
    "accept": "application/json",
    "Content-Type": "application/json",
}
data = {
    "request": {
        "variable": ["river_discharge"],
        "experiment": ["rcp_8_5"],
        "period": ["2041_2070"],
            "time_aggregation": "annual_mean",
    },
    "feature": {
        "type": "boundingbox",
        "coordinates": [[49.5, 5.5], [50.5, 6.5]],
    },
}

response = requests.post(url, headers=headers, json=data)

# Print the response
print(response.status_code)
print(response.json())
```

```python
from covjsonkit.api import Covjsonkit
decoder = Covjsonkit().decode(response.json())

ds = decoder.to_xarray()
ds
```

## Polygon query

```python
import requests

url = "https://polytope-dss.ecmwf.int/api/v1/dataset/sis-hydrology-variables-derived-projections/extract/?download_file=test.zip"
headers = {
    "accept": "application/json",
    "Content-Type": "application/json",
}
data = {
    "request": {
        "variable": ["river_discharge"],
        "experiment": ["rcp_8_5"],
        "period": ["2041_2070"],
            "time_aggregation": "annual_mean",
    },
    "feature": {
        "type": "polygon",
        "coordinates": [
            [49.5, 5.5],
            [50.5, 5.5],
            [50.5, 6.5],
            [49.5, 6.5],
            [49.5, 5.5],
        ],
    },
}

response = requests.post(url, headers=headers, json=data)

# Print the response
print(response.status_code)
print(response.json())
```

```python
from covjsonkit.api import Covjsonkit
decoder = Covjsonkit().decode(response.json())

ds = decoder.to_xarray()
ds
```
