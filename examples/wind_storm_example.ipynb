# European Wind Storm — sis-european-wind-storm-indicators

This notebook demonstrates how to query the **European Wind Storm Indicators** dataset
(`sis-european-wind-storm-indicators`) via the Polytope DSS API.

The dataset provides windstorm footprints (maximum 10 m wind gust) for historical
European windstorm events. Both **bounding-box** and **polygon** spatial queries are shown.

The example uses storm **Daria (25 January 1990)**, one of the most damaging North Sea
windstorms on record.

The response is a [CoverageJSON](https://covjson.org) `CoverageCollection` that can be
decoded with [`covjsonkit`](https://github.com/ecmwf/covjsonkit) and converted to an
`xarray.Dataset`.

> **Note:** The `variable` field for this dataset is a string (`"all"`), not a list.

## Bounding-box query

```python
import requests

url = "https://polytope-dss.ecmwf.int/api/v1/dataset/sis-european-wind-storm-indicators/extract/?download_file=test.zip"
headers = {
    "accept": "application/json",
    "Content-Type": "application/json",
}
data = {
    "request": {
        # variable is a plain string for this dataset, not a list
        "variable": "all",
    },
    "feature": {
        "type": "boundingbox",
        "coordinates": [[49.5, 5.5], [50.5, 6.5]],
    },
}

response = requests.post(url, headers=headers, json=data)

# Print the response
print(response.status_code)
print(response.json())
```

```python
from covjsonkit.api import Covjsonkit
decoder = Covjsonkit().decode(response.json())

ds = decoder.to_xarray()
ds
```

## Polygon query

```python
import requests

url = "https://polytope-dss.ecmwf.int/api/v1/dataset/sis-european-wind-storm-indicators/extract/?download_file=test.zip"
headers = {
    "accept": "application/json",
    "Content-Type": "application/json",
}
data = {
    "request": {
        # variable is a plain string for this dataset, not a list
        "variable": "all",
    },
    "feature": {
        "type": "polygon",
        "coordinates": [
            [49.5, 5.5],
            [50.0, 5.5],
            [50.0, 6.0],
            [49.5, 6.0],
            [49.5, 5.5],
        ],
    },
}

response = requests.post(url, headers=headers, json=data)

# Print the response
print(response.status_code)
print(response.json())
```

```python
from covjsonkit.api import Covjsonkit
decoder = Covjsonkit().decode(response.json())

ds = decoder.to_xarray()
ds
```
