# CORDEX — projections-cordex-domains-single-levels

This notebook demonstrates how to query the **CORDEX Europe** dataset
(`projections-cordex-domains-single-levels`) via the Polytope DSS API.

The dataset covers the CORDEX Europe domain on a rotated-pole grid at 0.11° resolution.
Both **bounding-box** and **polygon** spatial queries are shown.

The response is a [CoverageJSON](https://covjson.org) `CoverageCollection` that can be
decoded with [`covjsonkit`](https://github.com/ecmwf/covjsonkit) and converted to an
`xarray.Dataset`.

## Bounding-box query

```python
import requests

url = "https://polytope-dss.ecmwf.int/api/v1/dataset/projections-cordex-domains-single-levels/extract/?download_file=test.zip"
headers = {
    "accept": "application/json",
    "Content-Type": "application/json",
}
data = {
    "request": {
        "variable": ["mean_precipitation_flux"],
        "start_year": ["1951"],
        "end_year": ["1955"],
        "temporal_resolution": "daily_mean",
    },
    "feature": {
        "type": "boundingbox",
        "coordinates": [[0.0, 0.0], [1.0, 1.0]],
    },
}

response = requests.post(url, headers=headers, json=data)

# Print the response
print(response.status_code)
print(response.json())
```

```python
from covjsonkit.api import Covjsonkit
decoder = Covjsonkit().decode(response.json())

ds = decoder.to_xarray()
ds
```

## Polygon query

```python
import requests

url = "https://polytope-dss.ecmwf.int/api/v1/dataset/projections-cordex-domains-single-levels/extract/?download_file=test.zip"
headers = {
    "accept": "application/json",
    "Content-Type": "application/json",
}
data = {
    "request": {
        "variable": ["mean_precipitation_flux"],
        "start_year": ["1951"],
        "end_year": ["1955"],
        "temporal_resolution": "daily_mean",
    },
    "feature": {
        "type": "polygon",
        "coordinates": [
            [0.0, 0.0],
            [0.0, 0.3],
            [0.3, 0.3],
            [0.3, 0.0],
            [0.0, 0.0],
        ],
    },
}

response = requests.post(url, headers=headers, json=data)

# Print the response
print(response.status_code)
print(response.json())
```

```python
from covjsonkit.api import Covjsonkit
decoder = Covjsonkit().decode(response.json())

ds = decoder.to_xarray()
ds
```
