# Climate Indicators — sis-ecde-climate-indicators

This dataset provides European Climate Data Explorer (ECDE) climate indicators derived from CORDEX regional climate simulations. It includes variables such as total precipitation and mean temperature with various temporal aggregations. The API returns data in CoverageJSON format (`CoverageCollection`), which can be decoded and converted to an xarray Dataset using `covjsonkit`.

## Bounding-box query

```python
import requests

url = "https://polytope-dss.ecmwf.int/api/v1/dataset/sis-ecde-climate-indicators/extract/?download_file=test.zip"
headers = {
    "accept": "application/json",
    "Content-Type": "application/json",
}
data = {
    "request": {
        "variable": ["total_precipitation"],
        "temporal_aggregation": ["yearly"],
    },
    "feature": {
        "type": "boundingbox",
        "coordinates": [[49.5, 5.5], [50.5, 6.5]],
    },
}

response = requests.post(url, headers=headers, json=data)

# Print the response
print(response.status_code)
print(response.json())
```

```python
from covjsonkit.api import Covjsonkit
decoder = Covjsonkit().decode(response.json())

ds = decoder.to_xarray()
ds
```

## Polygon query

```python
import requests

url = "https://polytope-dss.ecmwf.int/api/v1/dataset/sis-ecde-climate-indicators/extract/?download_file=test.zip"
headers = {
    "accept": "application/json",
    "Content-Type": "application/json",
}
data = {
    "request": {
        "variable": ["mean_temperature"],
        "temporal_aggregation": ["seasonal"],
        #"gcm": ["mpi_esm_lr"],
        #"rcm": ["cclm4_8_17"],
    },
    "feature": {
        "type": "polygon",
        "coordinates": [
            [49.5, 5.5],
            [50.5, 5.5],
            [50.5, 6.5],
            [49.5, 6.5],
            [49.5, 5.5],
        ],
    },
}

response = requests.post(url, headers=headers, json=data)

# Print the response
print(response.status_code)
print(response.json())
```

```python
from covjsonkit.api import Covjsonkit
decoder = Covjsonkit().decode(response.json())

ds = decoder.to_xarray()
ds
```
