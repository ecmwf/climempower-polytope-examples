# EFAS Forecast — efas-forecast

This notebook demonstrates how to query the **European Flood Awareness System (EFAS)
Forecast** dataset (`efas-forecast`) via the Polytope DSS API.

EFAS provides short-to-medium range flood forecasts for Europe. The data is on a
Lambert Azimuthal Equal-Area (LAEA) projection grid. Both **bounding-box** and
**polygon** spatial queries are shown.

The response is a [CoverageJSON](https://covjson.org) `CoverageCollection`. Because
EFAS uses a non-standard projected grid, the `covjsonkit` decoder may not handle the
output directly — the raw JSON can be inspected or processed with custom code.

> **Note:** EFAS data is served from the EWDS (Early Warning Data Service) endpoint
> rather than the standard CDS. This is handled automatically by the API.

## Bounding-box query

```python
import requests

url = "https://polytope-dss.ecmwf.int/api/v1/dataset/efas-forecast/extract/?download_file=test.zip"
headers = {
    "accept": "application/json",
    "Content-Type": "application/json",
}
data = {
    "request": {
        "variable": ["river_discharge_in_the_last_6_hours"],
    },
    "feature": {
        "type": "boundingbox",
        "coordinates": [[49.5, 5.5], [50.5, 6.5]],
    },
}

response = requests.post(url, headers=headers, json=data)

# Print the response
print(response.status_code)
print(response.json())
```

```python
# Inspect the CoverageJSON structure
covjson = response.json()
print("Type:", covjson["type"])
print("Number of coverages:", len(covjson["coverages"]))

# Show the first coverage
first = covjson["coverages"][0]
print("\nFirst coverage ranges:", list(first["ranges"].keys()))
print("Number of values:", len(first["ranges"][list(first["ranges"].keys())[0]]["values"]))
```

## Polygon query

```python
import requests

url = "https://polytope-dss.ecmwf.int/api/v1/dataset/efas-forecast/extract/?download_file=test.zip"
headers = {
    "accept": "application/json",
    "Content-Type": "application/json",
}
data = {
    "request": {
        "variable": ["river_discharge_in_the_last_6_hours"],
    },
    "feature": {
        "type": "polygon",
        "coordinates": [
            [49.5, 5.5],
            [50.0, 5.5],
            [50.0, 6.0],
            [49.5, 6.0],
            [49.5, 5.5],
        ],
    },
}

response = requests.post(url, headers=headers, json=data)

# Print the response
print(response.status_code)
print(response.json())
```

```python
# Inspect the CoverageJSON structure
covjson = response.json()
print("Type:", covjson["type"])
print("Number of coverages:", len(covjson["coverages"]))

# Show the first coverage
first = covjson["coverages"][0]
print("\nFirst coverage ranges:", list(first["ranges"].keys()))
print("Number of values:", len(first["ranges"][list(first["ranges"].keys())[0]]["values"]))
```
