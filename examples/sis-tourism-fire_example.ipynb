# Tourism and Fire Danger Indicators — sis-tourism-fire-danger-indicators

This dataset provides fire danger and tourism climate indicators derived from CORDEX Europe simulations at 0.11° resolution. It includes variables such as the seasonal fire weather index under different climate scenarios and time periods. The API returns data in CoverageJSON format (`CoverageCollection`), which can be decoded and converted to an xarray Dataset using `covjsonkit`.

## Bounding-box query

```python
import requests

url = "https://polytope-dss.ecmwf.int/api/v1/dataset/sis-tourism-fire-danger-indicators/extract/?download_file=test.zip"
headers = {
    "accept": "application/json",
    "Content-Type": "application/json",
}
data = {
    "request": {
        "variable": ["seasonal_fire_weather_index"],
        "experiment": "rcp4_5",
        "period": ["2021_2025"],
        "time_aggregation": "seasonal_indicators",
    },
    "feature": {
        "type": "boundingbox",
        "coordinates": [[0.0, 0.0], [1.0, 1.0]],
    },
}

response = requests.post(url, headers=headers, json=data)

# Print the response
print(response.status_code)
print(response.json())
```

```python
from covjsonkit.api import Covjsonkit
decoder = Covjsonkit().decode(response.json())

ds = decoder.to_xarray()
ds
```

## Polygon query

```python
import requests

url = "https://polytope-dss.ecmwf.int/api/v1/dataset/sis-tourism-fire-danger-indicators/extract/?download_file=test.zip"
headers = {
    "accept": "application/json",
    "Content-Type": "application/json",
}
data = {
    "request": {
        "variable": ["seasonal_fire_weather_index"],
        "experiment": "rcp4_5",
        "period": ["2021_2025"],
        "time_aggregation": "seasonal_indicators",
    },
    "feature": {
        "type": "polygon",
        "coordinates": [
            [0.0, 0.0],
            [0.0, 1.0],
            [1.0, 1.0],
            [1.0, 0.0],
            [0.0, 0.0],
        ],
    },
}

response = requests.post(url, headers=headers, json=data)

# Print the response
print(response.status_code)
print(response.json())
```

```python
from covjsonkit.api import Covjsonkit
decoder = Covjsonkit().decode(response.json())

ds = decoder.to_xarray()
ds
```
